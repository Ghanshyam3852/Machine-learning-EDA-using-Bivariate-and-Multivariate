```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
titanic = pd.read_csv('titanic.csv')
```

```python
titanic.head()
```

```python
flights = sns.load_dataset('flights')
```

```python
flights.head()
```

```python
iris = sns.load_dataset('iris')
```

```python
iris.head()
```

## Scatterplot (Numerical - Numerical)

```python
plt.figure(figsize=(6,4))
sns.scatterplot(data =tips,x='total_bill',y='tip')
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(data =tips,x='total_bill',y='tip',hue ='sex')
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(data =tips,x='total_bill',y='tip',hue ='sex', style='smoker')
```

## Bar plot (Numerical - Categorical)

```python
plt.figure(figsize=(6,4))
sns.barplot(data= titanic,x='Pclass' , y = 'Age')
```

```python
plt.figure(figsize=(6,4))
sns.barplot(data= titanic,x='Pclass' , y = 'Age',hue = titanic['Sex'])
```

```python
plt.figure(figsize=(6,4))
sns.barplot(data= titanic,x='Pclass' , y = 'Fare',hue = titanic['Sex'])
```

## Boxplot (Numerical - Categorical)

```python
plt.figure(figsize=(6,4))
sns.boxplot(data= titanic,x='Sex' , y = 'Age')
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(data= titanic,x='Sex' , y = 'Age',hue = titanic['Survived'])
```

## Distplot (Numerical - Categorical)

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'] ,hist = False)
sns.distplot(titanic[titanic['Survived']==1]['Age'] , hist = False)
```

## Heatmap (Categorical - Categorical)

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived'] ))
```

```python
(titanic.groupby ('Embarked').mean(numeric_only=True)['Survived']*100)
```

```python
(titanic.groupby ('Pclass').mean(numeric_only=True)['Survived']*100)
```

## Clustermap (Categorical - Categorical)

```python
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```

## Pairplot (Collection of Scatterplot)

```python
sns.pairplot(iris)
```

```python
sns.pairplot(iris ,hue ='species')
```

## Lineplot (Numerical - Numerical)

```python
new = flights.groupby('year').sum(numeric_only=True).reset_index()
```

```python
new
```

```python
sns.lineplot(data =new, x='year',y = 'passengers')
```

```python
sns.clustermap(flights.pivot_table(values='passengers', index='month',columns='year'))
```
